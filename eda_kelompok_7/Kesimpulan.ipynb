Pertanyaan: Berapa nilai rata-rata, median, dan modus dari konsentrasi polutan Karbon Monoksida (CO) di udara?
Konsentrasi polutan Karbon Monoksida (CO) di udara memiliki rata-rata sebesar 18,82 µg/m³, nilai tengahnya 12 µg/m³, dan konsentrasi CO yang paling sering muncul adalah 11 µg/m³

Pertanyaan: Berapa rentang (range) serta nilai minimum dan maksimum dari konsentrasi partikel PM10?
Rentang konsentrasi PM10 sebesar 57 µg/m³, yaitu dari nilai minimum 3 hingga nilai maksimum 60 µg/m³.

Pertanyaan: Berapa nilai IQR (Interquartile Range) untuk polutan PM10 dan bagaimana interpretasi rentang data 50% pertengahannya?
Nilai IQR sebesar 10 µg/m³, sehingga 50% data konsentrasi PM10 berada pada rentang 13–23 µg/m³.

Temuan yang paling menarik adalah perbedaan mean dan median yang cukup jauh, terutama pada CO dan SO₂, menunjukkan bahwa data cenderung tidak simetris (skewed). CO cenderung miring ke kanan, sedangkan SO₂ cenderung miring ke kiri. Bahkan pada persebaran SO₂ terlihat terbagi menjadi 2. Dan juga pada variabel kategorik ditemukan bahwa polusi udara di Tangerang Selatan berada pada kondisi sedang.

Analisis ini memiliki keterbatasan berupa kurangnya periode pengamatan, sehingga belum menggambarkan kondisi kualitas udara dalam periode yang lebih panjang. Selain itu, beberapa baris pada data O3 sebagian kosong.
