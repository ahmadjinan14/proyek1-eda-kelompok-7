```python
import pandas as pd
df = pd.read_csv("tangerangrangrang.csv")
df.isna().sum()
df.duplicated().sum()
```

```python
import matplotlib.pyplot as plt

plt.hist(df["SO2"], bins=20)
plt.title("SO2")
plt.xlabel("SO2 dalam ISPU")
plt.ylabel("Frekuensi")
plt.show()
```

Histogram tersebut menampilkan bahwa kadar SO2 di wilayah Tangerang Selatan terbagi menjadi 2, ada yang rendah dan ada yang sedang.

```python
plt.hist(df["CO"], bins=20)
plt.title("CO")
plt.xlabel("CO dalam ISPU")
plt.ylabel("Frekuensi")
plt.show()
```

Histogram tersebut menampilkan kadar CO yang rata-rata masih pada kondisi baik (0-25), walaupun di beberapa daerah ada yang > 50.

```python
plt.boxplot(df["PM10"], orientation="horizontal")
plt.title("PM10")
plt.xlabel("PM10 dalam ISPU")
plt.ylabel("Frekuensi")
plt.show()
```

Pada boxplot tersebut, didapatkan bahwa kadar PM10 masih berada pada kondisi baik dan hanya sedikit daerah yang berada di atas 50

```python
plt.bar(df["Category"], df.groupby("Category")["Category"].transform("count"))
plt.title("Category")
plt.xlabel("Kategori polusi udara")
plt.ylabel("Frekuensi")
plt.show()
```

Dari semua data yang didapatkan sebelumnya, banyak daerah yang polusi udaranya berada pada tingkat sedang dan dilanjut dengan baik. Sedangkan daerah dengan polusi udara tidak sehat sangat sedikit di bawah 100.
