```python
import pandas as pd
df = pd.read_csv("tanggerang_baru.csv")
kolom_dipilih = ["CO", "PM10", "SO2"]

ringkasan = []
for col in kolom_dipilih:
    s = df[col]
    
    ringkasan.append({
        "Rata-rata (Mean)": round(s.mean(), 2),
        "Median": s.median(),
        "Modus": s.mode()[0] if not s.mode().empty else None,
        "Minimum": s.min(),
        "Maksimum": s.max(),
        "Jangkauan (Range)": s.max() - s.min(),
        "Varians": round(s.var(), 2),
        "Simpangan Baku (Std)": round(s.std(), 2),
        "Q1 (Kuartil 1)": s.quantile(0.25),
        "Q3 (Kuartil 3)": s.quantile(0.75),
        "IQR (Interquartile)": (s.quantile(0.25)) - (s.quantile(0.75)),
    })
tabel_ringkasan = pd.DataFrame(ringkasan)
display(tabel_ringkasan)
```
